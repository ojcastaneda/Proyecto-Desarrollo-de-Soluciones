```python
from sklearn.tree import DecisionTreeClassifier
from model.analyze import analyze_variables, check_bias, print_tree
from model.train import (
    cross_validation,
    grid_report,
    load_data,
    random_seed,
    test_report_umbral,
    train_unbiased_model,
)
from model.model import save_model
import random

random.seed(random_seed)
```

```python
train_dataset, test_dataset = load_data("../data/raw_data.csv")
bias_prone_variables = ["Sexo", "Grupo Etario"]
experiment = "final"
case_folder = f"./reports/{experiment}"


def model_path(output_folder: str):
    return f"{output_folder}/model"
```

```python
output_folder = f"{case_folder}/biased"
grid, summary = cross_validation(
    [("model", DecisionTreeClassifier())],
    train_dataset,
    {
        "model__splitter": ["random", "best"],
        "model__max_depth": [3, 5, 7],
        "model__min_samples_split": [2, 3, 4],
        "model__min_samples_leaf": [1, 2, 3],
    },
    include_all=True,
    include_bias=True,
)
summary
```

```python
model = grid_report(grid, test_dataset, experiment)
```

```python
analyze_variables(model, test_dataset, output_folder)
print_tree(model, test_dataset, output_folder)
```

```python
output_folder = f"{case_folder}/complete"
grid, summary = cross_validation(
    [("model", DecisionTreeClassifier())],
    train_dataset,
    {
        "model__splitter": ["random", "best"],
        "model__max_depth": [35, 40, 45],
        "model__min_samples_split": [2, 3, 4],
        "model__min_samples_leaf": [1, 2, 3],
    },
    include_all=True,
)
summary
```

```python
model = grid_report(grid, test_dataset, experiment)
save_model(model, model_path(output_folder))
```

```python
analyze_variables(model, test_dataset, output_folder)
print_tree(model, test_dataset, output_folder)
check_bias(model, train_dataset, test_dataset, bias_prone_variables)
```

```python
output_folder = f"{case_folder}/complete_unbiased"
model = train_unbiased_model(model, train_dataset, test_dataset, experiment)
save_model(model, model_path(output_folder))
analyze_variables(model, test_dataset, output_folder)
print_tree(model, test_dataset, output_folder)
check_bias(model, train_dataset, test_dataset, bias_prone_variables)
```

```python
output_folder = f"{case_folder}/simple"
grid, summary = cross_validation(
    [("model", DecisionTreeClassifier())],
    train_dataset,
    {
        "model__splitter": ["random", "best"],
        "model__max_depth": [3, 5, 8],
        "model__min_samples_split": [2, 3, 4],
        "model__min_samples_leaf": [4, 5, 6],
    },
)
summary
```

```python
model = grid_report(grid, test_dataset, experiment)
save_model(model, model_path(output_folder))
```

```python
analyze_variables(model, test_dataset, output_folder)
print_tree(model, test_dataset, output_folder)
check_bias(model, train_dataset, test_dataset, bias_prone_variables)
```

```python
output_folder = f"{case_folder}/simple_unbiased"
model = train_unbiased_model(model, train_dataset, test_dataset, experiment)
save_model(model, model_path(output_folder))
analyze_variables(model, test_dataset, output_folder)
print_tree(model, test_dataset, output_folder)
check_bias(model, train_dataset, test_dataset, bias_prone_variables)
```
